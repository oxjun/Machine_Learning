```python
import pandas as pd

df = pd.read_csv('Salary_Data.csv')
df.head()
```

```python
import numpy as np

X = df['YearsExperience']
y = df['Salary']

# Ubah bentuk array X menjadi 2D (karena SVR membutuhkan format 2D)
X = np.array(X).reshape(-1, 1)
```

```python
from sklearn.svm import SVR

# Membangun model SVR dengan kernel RBF, C, dan gamma
model = SVR(kernel='rbf', C=1e3, gamma=0.1)

# Melatih model
model.fit(X, y)
```

```python
import matplotlib.pyplot as plt

plt.scatter(X, y, color='black', label='Data Asli')
plt.plot(X, model.predict(X), color='red', label='Model SVR')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.legend()
plt.show()
```
