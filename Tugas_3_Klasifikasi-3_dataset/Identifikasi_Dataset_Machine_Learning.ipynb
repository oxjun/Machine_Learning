## Identifikasi Dataset Machine Learning
## Binary Classification, Multiclass Classification, dan Multilabel Classification

**Tujuan:** mengidentifikasi sembilan dataset dari Kaggle berdasarkan fitur (X), target/label (y), jumlah kelas, dan karakteristik tugas klasifikasinya.

> **Catatan:** Notebook ini berfokus pada identifikasi dan pemeriksaan struktur label, bukan membangun atau mengevaluasi model. Unduh dataset dari tautan Kaggle yang tercantum, lalu letakkan file CSV di folder `data/` sesuai nama yang ditentukan pada bagian masing-masing. Untuk dataset kompetisi, pengguna mungkin perlu login dan menyetujui aturan kompetisi terlebih dahulu.

### Perbedaan jenis klasifikasi
- **Binary classification:** satu target dengan tepat dua kelas.
- **Multiclass classification:** satu target dengan lebih dari dua kelas; setiap sampel hanya mempunyai satu kelas.
- **Multilabel classification:** satu sampel dapat mempunyai beberapa label aktif sekaligus. Biasanya target berupa beberapa kolom biner (0/1).



## 1. Persiapan 

```python
from pathlib import Path
import pandas as pd
import numpy as np

DATA_DIR = Path("data")
DATA_DIR.mkdir(exist_ok=True)

print("Folder data:", DATA_DIR.resolve())
print("Letakkan file CSV hasil unduhan Kaggle di folder data/.")
```

# A. Binary Classification

Pada klasifikasi biner, target hanya mempunyai dua kemungkinan kelas. Ketiga dataset berikut memiliki target biner yang jelas.

## A1. Pima Indians Diabetes Database

**Kaggle:** [Pima Indians Diabetes Database](https://www.kaggle.com/datasets/uciml/pima-indians-diabetes-database)

- **Tujuan:** memprediksi apakah pasien pada data mengalami diabetes.
- **Fitur (X):** `Pregnancies`, `Glucose`, `BloodPressure`, `SkinThickness`, `Insulin`, `BMI`, `DiabetesPedigreeFunction`, dan `Age`.
- **Target (y):** `Outcome`.
- **Label:** `0` = tidak terindikasi diabetes pada dataset; `1` = terindikasi diabetes.
- **Identifikasi:** binary classification karena `Outcome` memiliki dua nilai kelas, yaitu 0 dan 1.
- **Alasan dipilih:** fitur dan target mudah dikenali, sehingga cocok untuk latihan memahami pemisahan fitur-target dan klasifikasi biner.

**Catatan:** ini adalah dataset untuk pembelajaran, bukan alat diagnosis medis.

```python
# Ganti nama/path file sesuai file yang diunduh dari Kaggle.
diabetes_path = DATA_DIR / "diabetes.csv"

if diabetes_path.exists():
    diabetes = pd.read_csv(diabetes_path)
    print("Ukuran data:", diabetes.shape)
    print("Kolom:", diabetes.columns.tolist())
    print("\nDistribusi target Outcome:")
    print(diabetes["Outcome"].value_counts(dropna=False))
    print("\nJumlah kelas target:", diabetes["Outcome"].nunique())
else:
    print(f"File belum ditemukan: {diabetes_path}")
    print("Unduh CSV dari tautan Kaggle pada bagian A1.")
```

## A2. Breast Cancer Wisconsin (Diagnostic)

**Kaggle:** [Breast Cancer Wisconsin (Diagnostic)](https://www.kaggle.com/datasets/uciml/breast-cancer-wisconsin-data)

- **Tujuan:** mengklasifikasikan sampel tumor ke kategori jinak atau ganas.
- **Fitur (X):** pengukuran karakteristik inti sel, misalnya `radius_mean`, `texture_mean`, `perimeter_mean`, `area_mean`, dan fitur numerik lainnya.
- **Target (y):** `diagnosis`.
- **Label:** `B` = benign (jinak); `M` = malignant (ganas).
- **Identifikasi:** binary classification karena target `diagnosis` hanya mempunyai dua kategori.
- **Alasan dipilih:** labelnya eksplisit dan dataset menyediakan banyak fitur numerik untuk membedakan dua kelas.

**Catatan preprocessing:** kolom `id` bukan fitur prediktif; `Unnamed: 32` (jika ada) biasanya kosong dan perlu dibuang. Dataset ini untuk pembelajaran, bukan diagnosis klinis.

```python
breast_path = DATA_DIR / "data.csv"

if breast_path.exists():
    breast = pd.read_csv(breast_path)
    print("Ukuran data:", breast.shape)
    print("Kolom target diagnosis:")
    print(breast["diagnosis"].value_counts(dropna=False))
    print("Jumlah kelas:", breast["diagnosis"].nunique())
else:
    print(f"File belum ditemukan: {breast_path}")
    print("Unduh CSV dari tautan Kaggle pada bagian A2.")
```

## A3. Bank Marketing

**Kaggle:** [Bank Marketing Dataset](https://www.kaggle.com/datasets/janiobachmann/bank-marketing-dataset)

- **Tujuan:** memprediksi apakah nasabah akan berlangganan deposito berjangka setelah kampanye pemasaran.
- **Fitur (X):** antara lain `age`, `job`, `marital`, `education`, `balance`, `housing`, `loan`, dan informasi kontak/kampanye.
- **Target (y):** `deposit`.
- **Label:** `yes` = berlangganan; `no` = tidak berlangganan.
- **Identifikasi:** binary classification karena target `deposit` memiliki dua kategori.
- **Alasan dipilih:** mewakili kasus bisnis nyata dan mengandung fitur numerik maupun kategorikal, sehingga menarik untuk tahap preprocessing.

**Catatan:** jangan gunakan kolom target `deposit` sebagai fitur input.

```python
bank_path = DATA_DIR / "bank.csv"

if bank_path.exists():
    bank = pd.read_csv(bank_path)
    print("Ukuran data:", bank.shape)
    print("Distribusi target deposit:")
    print(bank["deposit"].value_counts(dropna=False))
    print("Jumlah kelas:", bank["deposit"].nunique())
else:
    print(f"File belum ditemukan: {bank_path}")
    print("Unduh CSV dari tautan Kaggle pada bagian A3.")
```

# B. Multiclass Classification

Pada klasifikasi multikelas, satu sampel hanya memiliki **satu** kelas target, tetapi pilihan kelasnya lebih dari dua.

## B1. Iris Species

**Kaggle:** [Iris Species](https://www.kaggle.com/datasets/uciml/iris)

- **Tujuan:** memprediksi spesies bunga Iris.
- **Fitur (X):** `SepalLengthCm`, `SepalWidthCm`, `PetalLengthCm`, dan `PetalWidthCm` (nama kolom dapat sedikit berbeda pada salinan dataset).
- **Target (y):** `Species`.
- **Kelas:** Iris-setosa, Iris-versicolor, dan Iris-virginica.
- **Identifikasi:** multiclass classification karena target memiliki tiga kelas dan setiap bunga hanya termasuk satu spesies.
- **Alasan dipilih:** dataset kecil, sederhana, dan sangat sesuai untuk demonstrasi klasifikasi multikelas.

```python
iris_path = DATA_DIR / "Iris.csv"

if iris_path.exists():
    iris = pd.read_csv(iris_path)
    target = "Species" if "Species" in iris.columns else "species"
    print("Ukuran data:", iris.shape)
    print(iris[target].value_counts(dropna=False))
    print("Jumlah kelas:", iris[target].nunique())
else:
    print(f"File belum ditemukan: {iris_path}")
    print("Unduh CSV dari tautan Kaggle pada bagian B1.")
```

## B2. Dry Bean Dataset

**Kaggle:** [Dry Bean Dataset](https://www.kaggle.com/datasets/muratkokludataset/dry-bean-dataset)

- **Tujuan:** mengidentifikasi jenis biji kacang kering berdasarkan ukuran dan bentuknya.
- **Fitur (X):** pengukuran seperti `Area`, `Perimeter`, `MajorAxisLength`, `MinorAxisLength`, `AspectRation`, dan karakteristik geometris lainnya.
- **Target (y):** `Class`.
- **Kelas:** BARBUNYA, BOMBAY, CALI, DERMASON, HOROZ, SEKER, dan SIRA.
- **Identifikasi:** multiclass classification karena terdapat tujuh kelas dan satu biji diberi satu kelas jenis kacang.
- **Alasan dipilih:** mempunyai lebih banyak kelas daripada Iris, sehingga memperlihatkan kasus klasifikasi multikelas yang lebih kompleks.

```python
drybean_path = DATA_DIR / "Dry_Bean_Dataset.xlsx"

if drybean_path.exists():
    drybean = pd.read_excel(drybean_path)
    print("Ukuran data:", drybean.shape)
    print(drybean["Class"].value_counts(dropna=False))
    print("Jumlah kelas:", drybean["Class"].nunique())
else:
    print(f"File belum ditemukan: {drybean_path}")
    print("Dataset Kaggle dapat berformat XLSX; jika CSV, gunakan pd.read_csv().")
```

## B3. Wine Recognition

**Kaggle:** [Wine Dataset](https://www.kaggle.com/datasets/harrywang/wine-dataset)

- **Tujuan:** mengklasifikasikan sampel wine ke salah satu dari tiga kelompok/kultivar.
- **Fitur (X):** karakteristik kimia, misalnya alcohol, malic acid, ash, alcalinity, magnesium, dan komponen kimia lainnya.
- **Target (y):** kolom kelas/jenis wine (nama kolom dapat berbeda menurut versi file).
- **Kelas:** tiga kelompok wine (umumnya dikodekan 1, 2, dan 3 pada Wine Recognition dataset).
- **Identifikasi:** multiclass classification apabila file yang diunduh memuat satu kolom target dengan tepat tiga kelas.
- **Alasan dipilih:** menunjukkan penggunaan fitur kimia numerik untuk membedakan lebih dari dua kategori.

**Pemeriksaan wajib:** pastikan file Kaggle yang diunduh benar-benar merupakan Wine Recognition dengan tiga kelas. Jangan tertukar dengan dataset Wine Quality yang targetnya berupa skor kualitas.

```python
wine_path = DATA_DIR / "wine.csv"

if wine_path.exists():
    wine = pd.read_csv(wine_path)
    print("Ukuran data:", wine.shape)
    print("Kolom:", wine.columns.tolist())
    # Periksa nama kolom target pada file yang diunduh.
    possible_targets = [c for c in wine.columns if c.lower() in {"class", "target", "cultivar", "wine"}]
    print("Kandidat kolom target:", possible_targets)
    if possible_targets:
        target = possible_targets[0]
        print(wine[target].value_counts(dropna=False))
        print("Jumlah kelas:", wine[target].nunique())
    else:
        print("Tentukan kolom target sesuai dokumentasi dataset.")
else:
    print(f"File belum ditemukan: {wine_path}")
    print("Unduh CSV dari tautan Kaggle pada bagian B3.")
```

# C. Multilabel Classification

Pada klasifikasi multilabel, satu sampel dapat memiliki beberapa label aktif sekaligus. Target biasanya terdiri dari beberapa kolom label bernilai 0/1. Hal ini berbeda dari multiclass, yang hanya memilih satu kelas untuk setiap sampel.

## C1. Jigsaw Toxic Comment Classification Challenge

**Kaggle:** [Toxic Comment Classification Challenge](https://www.kaggle.com/competitions/jigsaw-toxic-comment-classification-challenge/data)

- **Tujuan:** mendeteksi jenis perilaku berbahaya pada komentar.
- **Fitur (X):** `comment_text`.
- **Target (Y):** `toxic`, `severe_toxic`, `obscene`, `threat`, `insult`, dan `identity_hate`.
- **Nilai label:** 0 = label tidak berlaku; 1 = label berlaku.
- **Identifikasi:** multilabel classification karena satu komentar dapat memiliki beberapa label bernilai 1 sekaligus.
- **Contoh:** komentar dapat diberi label `toxic=1`, `obscene=1`, dan `insult=1` pada waktu yang sama.
- **Alasan dipilih:** contoh multilabel yang sangat jelas, memiliki enam kolom target biner, dan menggambarkan penerapan klasifikasi teks.

**File utama:** biasanya `train.csv`. File kompetisi dapat memerlukan persetujuan aturan Kaggle.

```python
toxic_path = DATA_DIR / "toxic_train.csv"
toxic_labels = ["toxic", "severe_toxic", "obscene", "threat", "insult", "identity_hate"]

if toxic_path.exists():
    toxic = pd.read_csv(toxic_path)
    print("Ukuran data:", toxic.shape)
    print("Kolom label yang tersedia:", [c for c in toxic_labels if c in toxic.columns])
    print("\nJumlah komentar per label:")
    print(toxic[toxic_labels].sum())
    print("\nContoh jumlah label aktif per komentar:")
    print(toxic[toxic_labels].sum(axis=1).value_counts().sort_index())
else:
    print(f"File belum ditemukan: {toxic_path}")
    print("Unduh train.csv dari kompetisi Kaggle dan simpan sebagai toxic_train.csv.")
```

## C2. Multi-label Image Classification (COMP5329)

**Kaggle:** [COMP5329 Multi-label Classification Competition](https://www.kaggle.com/competitions/comp5329s1-2023)

- **Tujuan:** memprediksi beberapa kategori yang muncul dalam satu gambar.
- **Fitur (X):** gambar; kompetisi juga dapat menyediakan caption/teks pendamping.
- **Target (Y):** daftar label kategori untuk setiap gambar (format label mengikuti file kompetisi).
- **Identifikasi:** multilabel classification jika satu baris data dapat memuat lebih dari satu label untuk gambar yang sama.
- **Contoh konsep:** satu gambar dapat memuat beberapa objek/kategori, sehingga lebih dari satu label benar.
- **Alasan dipilih:** memperlihatkan bahwa multilabel tidak hanya diterapkan pada teks, tetapi juga pada computer vision.

**Catatan penting:** periksa format kolom label dan aturan kompetisi pada file yang diunduh. Label dapat disimpan sebagai daftar angka dalam satu kolom, bukan selalu sebagai enam atau lebih kolom biner.

```python
# Setelah mengunduh file anotasi/CSV kompetisi, sesuaikan nama file dan kolomnya.
image_labels_path = DATA_DIR / "comp5329_train.csv"

if image_labels_path.exists():
    image_labels = pd.read_csv(image_labels_path)
    print("Ukuran data:", image_labels.shape)
    print("Kolom:", image_labels.columns.tolist())
    display(image_labels.head())
    print("\nPeriksa kolom label pada dokumentasi kompetisi.")
else:
    print(f"File belum ditemukan: {image_labels_path}")
    print("Unduh data kompetisi dan simpan file anotasi CSV di folder data/.")
```

## C3. Movie Genre Multimodal (MM-IMDb)

**Kaggle:** [Movie Genre Multimodal](https://www.kaggle.com/competitions/joai-movie-genre-multimodal/overview)

- **Tujuan:** memprediksi genre sebuah film.

- **Target (Y):** genre film.
- **Label:** satu film dapat memiliki beberapa genre, misalnya Action, Adventure, dan Sci-Fi.
- **Identifikasi:** multilabel classification karena satu film dapat memiliki lebih dari satu genre secara bersamaan.
- **Alasan dipilih:** contoh yang mudah dipahami karena film sering masuk ke beberapa genre sekaligus; juga memperlihatkan penggunaan data teks dan gambar.

**Catatan:** pastikan berkas yang diunduh benar-benar menyediakan beberapa genre untuk satu film. Format label dan daftar genre mengikuti versi dataset/kompetisi.

```python
movie_path = DATA_DIR / "movie_genres.csv"

if movie_path.exists():
    movies = pd.read_csv(movie_path)
    print("Ukuran data:", movies.shape)
    print("Kolom:", movies.columns.tolist())
    display(movies.head())
    print("\nPeriksa kolom genre: apakah satu film dapat memiliki beberapa genre?")
else:
    print(f"File belum ditemukan: {movie_path}")
    print("Unduh data dari Kaggle dan simpan metadata/label sebagai movie_genres.csv.")
```

## 3. Kesimpulan

Binary classification digunakan ketika target hanya mempunyai dua kelas. Multiclass classification digunakan ketika target mempunyai lebih dari dua kelas, tetapi setiap sampel hanya memiliki satu kelas. Multilabel classification digunakan ketika satu sampel dapat mempunyai beberapa label sekaligus.

Sembilan dataset dipilih karena mewakili contoh penerapan yang berbeda: kesehatan dan pemasaran untuk binary, pengenalan spesies/jenis untuk multiclass, serta moderasi komentar, pengenalan gambar, dan genre film untuk multilabel.
